# WaTEM/SEDEM Output Analysis: Python-Based Analysis of Catchment Sediment Results
Objective
The aim of this analysis is to extract and evaluate catchment-scale sediment results generated by the WaTEM/SEDEM model.
The analysis focuses on:
- Total soil erosion
- Total sediment deposition
- Deposition and export ratios

## Reading the Model Output

```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt

# Define the file path
file_path = "data/Total sediment.txt"

# Read the output file
with open(file_path, "r") as file:
    content = file.read()

# Display the file content
print(content)
```

```python
# Split the content into individual lines
lines = content.splitlines()

# Create an empty dictionary
results = {}

# Extract numerical values
for line in lines:
    if ":" in line:
        key, value = line.split(":")
        value = value.split("(")
        value = value[0]
        value = value.strip()
        results[key] = float(value)

# Display extracted results
print(results)
```

### 2. Creating a Summary Table

```python
# Create a summary table
df = pd.DataFrame(results.items(), columns=["Parameter", "Value"])
print(df)
```

### 3. Calculating the Deposition Ratio
The deposition ratio represents the relationship between total sediment deposition and the absolute value of total soil erosion.

```python
# Calculate deposition ratio (%)
total_erosion = abs(results["Total erosion"])
total_deposition = results["Total deposition"]
deposition_ratio = (total_deposition / total_erosion) * 100
print("Deposition ratio:", deposition_ratio, "%")
```

### 4. Calculating the Sediment Export Ratio
The sediment export ratio represents the percentage of total soil erosion that leaves the catchment.

```python
# Calculate sediment export ratio (%)
total_export = (results["Sediment leaving the catchment, via the river"]+ results["Sediment leaving the catchment, not via the river"])
export_ratio = (total_export / total_erosion) * 100
print("Total sediment export ratio is:", export_ratio, "%")
```

**Note:** The deposition ratio and the export ratio do not add up to 100%. Sediment entering the sewer system and sediment trapped in buffers
are reported separately in the model output and are not included in these two ratios.

### 5. Summary of Calculated Results

```python
# Create a summary table of the calculated results
summary = {"Total erosion (kg)":results["Total erosion"],
           "Total deposition (kg)":total_deposition,
           "Deposition ratio (%)":deposition_ratio,
           "Export ratio (%)": export_ratio}
summary_df= pd.DataFrame(summary.items(), columns = ["Parameter","Value"])
print("Summary table is:",summary_df)
```
